**Título:** Árvores Binárias de Busca Balanceadas (AVL) e Rotações

> **Enunciado Original:**
> Projetar a estrutura de dados de uma Árvore AVL para indexação de sensores em tempo real. Demonstrar formalmente as rotações simples à esquerda/direita e dupla esquerda-direita para restabelecer o fator de balanceamento.

### Formulação Matemática Formal
$$FB(v) = h(v.esq) - h(v.dir), \quad |FB(v)| \le 1$$


### Passo 1: Condição de Balanceamento AVL
Um nó v está balanceado se seu fator de balanceamento FB(v) pertencer ao conjunto \{-1, 0, +1\}.


### Passo 2: Detecção de Desbalanceamento
Após a inserção de chaves consecutivas crescentes, a subárvore direita atinge altura h + 2, gerando FB(v) = -2 (caso Direita-Direita).


### Passo 3: Aplicação da Rotação Simples à Esquerda
O nó filho direito y torna-se a nova raiz da subárvore, com o nó original v tornando-se filho esquerdo de y.


### Passo 4: Atualização das Alturas
h_{nova}(v) = 1 + \max(h(v.esq), h(v.dir)), restabelecendo a altura máxima da árvore em O(\log n).


### Passo 5: Conclusão Estrutural
A indexação de 10.000 sensores de campo é garantida em no máximo \lceil 1.44 \log_2(10.000) \rceil = 19 comparações de ponteiros.


```python
import numpy as np
import matplotlib.pyplot as plt

nodes = np.arange(1, 1000)
h_unbal = nodes
h_avl = 1.44 * np.log2(nodes)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0f172a')
ax.set_facecolor('#0f172a')
ax.plot(nodes, h_unbal, color='#ef4444', lw=2, label='BST Degenerada O(n)')
ax.plot(nodes, h_avl, color='#38bdf8', lw=2.5, label='Árvore AVL O(log n)')
ax.set_title('Altura da Árvore vs Número de Nós', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Quantidade de Nós (N)', color='#94a3b8')
ax.set_ylabel('Altura Máxima (Ponteiros)', color='#94a3b8')
ax.set_yscale('log')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#2e3b52', alpha=0.5)
ax.legend(facecolor='#1e293b', edgecolor='#38bdf8', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
