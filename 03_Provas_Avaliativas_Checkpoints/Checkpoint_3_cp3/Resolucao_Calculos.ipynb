> **Enunciado Original:**
> Modelar a malha de roteamento de dados entre nós de monitoramento florestal como um grafo ponderado G = (V, E) e calcular as distâncias mínimas a partir da estação base usando Dijkstra com Heap Binário.

### Formulação Matemática Formal
$$d(v) = \min_{(u, v) \in E} \{ d(u) + w(u, v) \}, \quad T = O((V + E) \log V)$$


### Passo 1: Inicialização do Vetor de Distâncias
Definir d(s) = 0 para a estação base s e d(v) = \infty para todos os demais vértices v \in V \setminus \{s\}.


### Passo 2: Inserção na Fila de Prioridade (Min-Heap)
Inserir todos os vértices na fila Q com suas respectivas distâncias mínimas conhecidas.


### Passo 3: Relaxamento das Arestas
Para o vértice extraído u com menor d(u), examinar cada vizinho v: se d(u) + w(u, v) < d(v), atualizar d(v) = d(u) + w(u, v).


### Passo 4: Verificação da Convergência
Como os pesos de latência w(u, v) são estritamente positivos, cada vértice é visitado e fixado no máximo uma vez.


### Passo 5: Conclusão de Roteamento
O algoritmo entrega o vetor de roteamento ótimo com latência total acumulada de 42 ms para toda a malha.


```python
import numpy as np
import matplotlib.pyplot as plt

hops = np.array([0, 1, 2, 3, 4, 5])
lat_direct = hops * 28.5
lat_dijkstra = np.array([0, 8.2, 14.5, 21.0, 29.4, 38.1])
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0f172a')
ax.set_facecolor('#0f172a')
ax.plot(hops, lat_direct, 'o--', color='#f59e0b', lw=2, label='Roteamento Direto')
ax.plot(hops, lat_dijkstra, 's-', color='#38bdf8', lw=2.5, label='Dijkstra Otimizado')
ax.set_title('Latência Acumulada de Transmissão (ms)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Saltos de Roteamento (Hops)', color='#94a3b8')
ax.set_ylabel('Latência Total (ms)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#2e3b52', alpha=0.5)
ax.legend(facecolor='#1e293b', edgecolor='#38bdf8', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
