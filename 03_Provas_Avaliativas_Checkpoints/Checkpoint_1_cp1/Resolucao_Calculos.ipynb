> **Enunciado Original:**
> Implementar e comparar o desempenho temporal e espacial dos algoritmos de ordenação QuickSort e MergeSort frente ao InsertionSort para vetores de tamanhos N = [1.000, 10.000, 50.000, 100.000] inteiros aleatórios.

### Formulação Matemática Formal
$$T(n) = 2T(n/2) + \Theta(n) \implies T(n) = \Theta(n \log n)$$


### Passo 1: Definição da Recorrência
A relação de recorrência para o MergeSort divide o vetor em 2 metades e intercala linearmente: T(n) = 2T(n/2) + cn


### Passo 2: Aplicação do Teorema Mestre
Temos a = 2, b = 2 e f(n) = cn. Como \log_b(a) = \log_2(2) = 1, e f(n) = \Theta(n^1), estamos no Caso 2 do Teorema Mestre.


### Passo 3: Solução Fechada da Complexidade
T(n) = \Theta(n^{\log_b a} \log n) = \Theta(n \log n). O número total de operações para n = 100.000 é \approx 1,66 \times 10^6.


### Passo 4: Comparativo com Algoritmos Quadráticos
Para o InsertionSort com T(n) = \Theta(n^2), para n = 100.000 temos 10^{10} operações, representando uma aceleração teórica de mais de 6.000 vezes.


### Passo 5: Conclusão de Eficiência
Para conjuntos massivos de dados no ambiente Natura de telemetria, o algoritmo MergeSort assegura estabilidade temporal determinística O(n \log n).


```python
import numpy as np
import matplotlib.pyplot as plt

n = np.linspace(100, 10000, 100)
t_quad = (n**2) / 1e6
t_nlogn = (n * np.log2(n)) / 1e4
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0f172a')
ax.set_facecolor('#0f172a')
ax.plot(n, t_quad, color='#ef4444', lw=2, label='InsertionSort O(n²)')
ax.plot(n, t_nlogn, color='#38bdf8', lw=2.5, label='MergeSort O(n log n)')
ax.set_title('Comparativo de Eficiência Assintótica', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tamanho da Entrada (N)', color='#94a3b8')
ax.set_ylabel('Tempo Normalizado (ms)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#2e3b52', alpha=0.5)
ax.legend(facecolor='#1e293b', edgecolor='#38bdf8', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
